# 다익스트라 알고리즘

## 1. 그래프의 기본 개념
- **정점(Vertex)**: 도시나 장소와 같은 지점
- **간선(Edge)**: 정점과 정점을 연결하는 선
- **가중치(Weight)**: 간선에 부여된 거리, 시간, 비용 등의 수치

#### 예시:
```
A --(5)-- B
```
A에서 B로 연결된 간선의 가중치는 5이다.


## 2. 우선순위 큐(힙)의 개념

`만약 최단 경로 문제에서 전체 노드의 개수가 5,000개 이하라면 이코드로 문제를 해결할 수 있지만, 노드의 개수가 10,000개 넘어가는 문제라면?`
<br> 
`-> 1억번의 반복을 해야하므로, 더 빠른 최단 경로를 찾기 위해 우선순위 큐(힙)을 사용하게 되는 것이다.`

- **가장 우선순위가 높은** 데이터를 빠르게 꺼낼 수 있는 자료구조이다.
- 다익스트라 알고리즘에서는 **최소 힙(min-heap)** 을 사용한다.
- 파이썬에서는 heapq 모듈로 구현할 수 있다.

```python
import heapq
queue = []
heapq.heappush(queue, (2, 'A'))
heapq.heappush(queue, (1, 'B'))
print(heapq.heappop(queue))  # (1, 'B')
```

## 3. 다익스트라 알고리즘 소개 및 전제 조건
- 하나의 정점에서 다른 모든 정점까지의 최단 경로를 계산한다.
- 모든 간선의 **가중치는 0 또는 양수**여야 한다.
- 음수 가중치가 있는 경우에는 벨만-포드 알고리즘을 사용해야 한다.


## 4. 알고리즘 동작 원리 요약 (표 형식)

| 단계 | 설명 | 코드 |
|------|------|------|
| **1** | **모든 정점의 거리를 무한대로 초기화하고 시작 정점은 0으로 설정한다.**<br>📌 처음에는 어디로 가야 할지 모르기 때문에, 출발점을 제외한 모든 거리는 무한대로 설정한다. | `distances = {node: float('inf') for node in graph}` <br> `distances[start] = 0` |
| **2** | **시작 정점을 우선순위 큐에 넣는다.**<br>📌 출발 지점을 큐에 넣고, 거리 탐색을 시작할 준비를 한다. | `queue = [(0, start)]` |
| **3** | **큐에서 가장 짧은 거리의 정점을 꺼낸다.**<br>📌 현재까지의 거리 기준으로 가장 가까운 지점을 꺼내어 탐색을 시작한다. | `current_distance, current_node = heapq.heappop(queue)` |
| **4** | **이미 처리된 정점이면 건너뛴다.**<br>📌 이미 더 짧은 거리로 방문했던 지점이라면 다시 계산하지 않고 건너뛴다. | `if current_distance > distances[current_node]:` <br>   `continue` |
| **5** | **인접 정점을 확인하고 더 짧은 경로가 있다면 거리 정보를 갱신한다.**<br>📌 현재 지점에서 갈 수 있는 이웃을 살펴보고, 더 짧게 갈 수 있는 경로가 있는지 확인한다. | `for neighbor, weight in graph[current_node]:`<br> `distance = current_distance + weight` |
| **6** | **갱신된 정점을 큐에 삽입한다.**<br>📌 더 짧은 거리로 갈 수 있음을 확인했으면, 해당 지점을 큐에 넣어 다음에 확인할 수 있게 한다. | `if distance < distances[neighbor]:` <br>   `distances[neighbor] = distance`<br>    `heapq.heappush(queue, (distance, neighbor))` |
| **7** | **큐가 빌 때까지 또는 목표 정점을 꺼낼 때까지 반복한다.**<br>📌 탐색할 지점이 남아 있는 동안에는 위의 과정을 계속 반복한다. | `while queue:`<br>    `# 반복 수행` |

## 5. 탐욕적/동적 프로그래밍적 성격
- **탐욕적(Greedy)**: 현재 가장 가까운 정점을 선택한다.
- **동적 프로그래밍(DP)**: 거리 테이블을 반복적으로 갱신한다.


## 6. 파이썬 코드 구현

````python
import heapq

def dijkstra(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    queue = [(0, start)]

    while queue:
        current_distance, current_node = heapq.heappop(queue)

        if current_distance > distances[current_node]:
            continue

        for neighbor, weight in graph[current_node]:
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(queue, (distance, neighbor))

    return distances

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('C', 2), ('D', 5)],
    'C': [('D', 1)],
    'D': []
}

print(dijkstra(graph, 'A'))
````

- 해석 구문

````python
import heapq  # heapq는 '우선순위 큐(priority queue)'를 사용하기 위한 파이썬 내장 모듈입니다.
              # 이 알고리즘에서는 가장 짧은 거리를 가진 노드를 빠르게 꺼내기 위해 사용합니다.

# 다익스트라 알고리즘 정의: 특정 시작점(start)에서 모든 노드까지의 최단거리를 계산합니다.
def dijkstra(graph, start):
    # 모든 노드의 거리를 무한대로 초기화합니다. (아직 갈 수 없다고 가정)
    distances = {node: float('inf') for node in graph}
    distances[start] = 0  # 시작 노드는 거리가 0입니다. (자기 자신이므로)

    # 우선순위 큐(힙)에 (거리, 노드) 형태로 초기값 저장. 시작점부터 출발합니다.
    queue = [(0, start)]

    # 큐가 빌 때까지 반복합니다.
    while queue:
        # 현재 큐에서 가장 짧은 거리를 가진 노드를 꺼냅니다.
        current_distance, current_node = heapq.heappop(queue)

        # 만약 지금 꺼낸 거리 정보가 이미 알고 있는 거리보다 크면, 무시합니다.
        if current_distance > distances[current_node]:
            continue  # 더 좋은 경로가 이미 발견된 상태이므로 건너뜁니다.

        # 현재 노드와 연결된 이웃 노드들을 확인합니다.
        for neighbor, weight in graph[current_node]:
            # 현재 노드를 거쳐서 이웃 노드로 가는 거리 계산
            distance = current_distance + weight
            # 만약 이 경로가 지금까지 알려진 거리보다 짧다면 업데이트합니다.
            if distance < distances[neighbor]:
                distances[neighbor] = distance  # 더 짧은 거리로 갱신
                heapq.heappush(queue, (distance, neighbor))  # 큐에 새로운 경로 추가

    # 모든 노드까지의 최단 거리 정보가 담긴 딕셔너리를 반환합니다.
    return distances

# 그래프 구조를 딕셔너리로 정의합니다.
# 각 노드는 연결된 노드와 그 노드까지의 거리(weight)를 리스트로 가집니다.
graph = {
    'A': [('B', 1), ('C', 4)],  # A는 B까지 1, C까지 4의 거리로 연결됨
    'B': [('C', 2), ('D', 5)],  # B는 C까지 2, D까지 5
    'C': [('D', 1)],            # C는 D까지 1
    'D': []                     # D는 더 이상 연결된 노드 없음 (종착점)
}

# A 노드를 시작점으로 하여 각 노드까지의 최단 거리를 출력합니다.
print(dijkstra(graph, 'A'))
````

## 7. 다른 알고리즘과의 비교

## 9. 연습 문제

다음 그래프에서 A부터 D까지의 최단 거리를 구하시오.
```
A → B (2), A → C (5)
B → C (1), B → D (4)
C → D (1)
```

A → B → C → D 경로가 최단이며, 총 거리는 4이다.


````python
import heapq  # 가장 짧은 거리의 노드를 빠르게 꺼내기 위한 모듈입니다 (우선순위 큐)

# 다익스트라 알고리즘 함수 정의
def dijkstra(graph, start):
    # 모든 노드의 거리를 무한대로 초기화 (처음엔 갈 수 없다고 가정)
    distances = {node: float('inf') for node in graph}
    distances[start] = 0  # 시작 노드는 자기 자신이므로 거리 0

    # 시작 노드를 큐에 넣습니다 (거리, 노드)
    queue = [(0, start)]

    # 큐가 빌 때까지 반복
    while queue:
        # 현재 가장 짧은 거리의 노드를 꺼냅니다
        current_distance, current_node = heapq.heappop(queue)

        # 만약 이미 더 짧은 거리로 방문한 적 있다면 건너뜁니다
        if current_distance > distances[current_node]:
            continue

        # 현재 노드와 연결된 이웃 노드들을 확인합니다
        for neighbor, weight in graph[current_node]:
            # 현재 거리 + 이웃 노드까지 거리
            distance = current_distance + weight
            # 더 짧은 경로를 찾았을 경우 갱신합니다
            if distance < distances[neighbor]:
                distances[neighbor] = distance  # 최단 거리 업데이트
                heapq.heappush(queue, (distance, neighbor))  # 큐에 넣기

    # 각 노드까지의 최단 거리를 담은 딕셔너리를 반환합니다
    return distances

# 문제에서 주어진 그래프를 딕셔너리로 표현
graph = {
    'A': [('B', 2), ('C', 5)],  # A는 B까지 2, C까지 5
    'B': [('C', 1), ('D', 4)],  # B는 C까지 1, D까지 4
    'C': [('D', 1)],            # C는 D까지 1
    'D': []                     # D는 연결된 노드 없음
}

# A에서 시작해서 각 노드까지의 최단 거리 계산
result = dijkstra(graph, 'A')

# D까지의 최단 거리 출력
print("A부터 D까지 최단 거리:", result['D']) # A부터 D까지 최단 거리: 4

````
