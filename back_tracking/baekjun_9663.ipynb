```python
# N-Queens 퀸문제
# N X N 크기의 정사각형 에 N개의 퀸이 서로 겹치지 않은 곳에 존재 해야 한다.
# 1. 열, 행 기준으로 1개씩만 존재해야 한다. 
# 2. 대각의 경우 우상향시 (x+y 의값이 모두 같음),
# 3. 우하향시 x=y

N = int(input())
ans = 0 # 개수 세기

v1 = [0] * N # 행과 열의 기준을 생각했을떄 1차원 배열로 만들어서 모든 인덱스에 값이 들어가게해서 1로 변환되는 리스트
v2 = [0] * (2*N-1)  # 우상향에 대한   
v3 = [0] * (2*N-1)  # 우하향에 대한

# n: 퀸의 시작 위치
def dfs(n):
    global ans
    
    if n==N:
        ans +=1
        return
    
    # 문제 풀기 전 생각
    # 1. 열, 행 기준으로 1개씩만 존재해야 한다.
    # 2. 대각의 경우 우상향시 (x+y 의값이 모두 같음),
    # 3. 우하향시 x=y | x-y=0
    # -----
    # 문제 푼 후 생각
    # 1. Q가 한 지점에 위치하면 행과 열의 인덱스는 같기에 하나의 1차원리스트로 인덱스표현해서 영역 기록 가능
    # 2. 우상향시 x+y의 값이 같은 방향선을 토대로 이(=x+y) 값을 인덱스로 반환해서 표현 -> x+y의 값이 같은 곳은 이제 퀸을 못둔다
    # 3. 우하향시 x-y의 값이 같은 방향선을 토대로 이(=x-y) 값을 인덱스로 반환해서 표현 -> x-y의 값이 같은 곳은 이제 퀸을 못둔다
    for i in range(N):
        if v1[i]==0 and v2[n+i]==0 and v3[n-i]==0:
            v1[i] = v2[n+i] = v3[n-i] = 1
            dfs(n+1) # (2,2)Q 들어간 상태에서 다시 재귀
            v1[i] = v2[n+i] = v3[n-i] = 0   #이전 상태로 돌아가서 다른 곳의 탐색도 진행해야 하기에 0으로 되돌림

dfs(0)
print(ans)
```
