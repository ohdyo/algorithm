```python
# 출력부
graph = [
  [1, 4], # 0번쨰 노드와 연결된 다른 노드
  [0, 2, 3, 4], # 1번쨰 노드와 연결된 다른 노드
  [1,3,4], # 2번쨰 노드와 연결된 다른 노드
  [1,2,4], # 3번쨰 노드와 연결된 다른 노드
  [0,1,3] # 4번쨰 노드와 연결된 다른 노드
]

from collections import deque

def bfs (graph, node):
# 큐 노드 구현 및 첫 node 큐에 삽입
    queue = deque([node])
    #첫 node 방문 처리
    visited = set() # or set()
    visited.add(node)

    #큐가 완전히 빌 때 까지 반복
    while queue:
        #큐에 삽입된 노드 순서대로 꺼내는 메서드popleft
        v = queue.popleft()
        #현재 처리 중인 노드에서 방문하지 않은 노드를 모두 큐에 삽입
        for i in graph[v]:
            if i not in visited:
                queue.append(i)     #1. queue: 1,4 | 2. queue: 4,0,2,3,4  | 3. queue: 3,4,1,3,4   | 4. queue:  4,1,3,4,2,1,2,4
                visited.add(i)   #1, visited 0,1,4  | 2. visited 0,1,4   | 3. visited : 0,1,4,2  | 4. visited : 0,1,4,2,3  
    return visited      

bfs(graph, 0)
```
