### 1. 버블 정렬 (Bubble Sort)
* 기본 아이디어:
버블 정렬은 리스트에서 인접한 두 원소를 비교하고, 크기가 잘못된 순서라면 두 원소를 교환하는 방식으로 리스트를 정렬하는 알고리즘이다. 
이 과정은 리스트 끝까지 반복되며, 매 반복 후 가장 큰 값이 리스트의 끝으로 "버블"처럼 올라간다.

* 단계:
    - 비교 및 교환: 리스트의 첫 번째 원소와 두 번째 원소를 비교한다. 만약 첫 번째 원소가 두 번째 원소보다 크면 두 원소를 교환한다. 이러한 비교와 교환을 리스트의 끝까지 반복한다.
    - 반복: 첫 번째 패스(전체 리스트를 한 번 반복) 후 가장 큰 원소가 맨 뒤로 이동한다. 이를 바탕으로 리스트의 끝을 제외하고 나머지 부분에서 다시 동일한 비교와 교환을 반복한다. 이 과정은 리스트의 크기가 1이 될 때까지 반복된다.

* 장점:
    - 단순함: 알고리즘이 매우 직관적이고 구현이 간단하다.
    - 메모리 사용 적음: 추가적인 공간을 사용하지 않으므로 메모리 사용이 적다. (in-place 정렬)

* 단점:
    - 비효율적: 시간 복잡도가 O(n²)로, 정렬해야 할 데이터가 많을수록 비효율적이다. 최악의 경우에도 매번 비교와 교환을 해야 하므로 속도가 느리다.
    - 최적화가 부족: 버블 정렬은 이미 정렬된 데이터를 처리하는 데 있어서 최적화가 부족하다. 데이터를 한번 순회하고 다시 반복하는 방식이므로 이미 정렬된 상태에서도 불필요하게 반복이 이루어진다.

* 시간 복잡도:
    - 최선: O(n) (이미 정렬된 경우, 한 번의 비교 후 종료)
    - 평균: O(n²) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n²) (리스트가 역순일 경우)

* 공간 복잡도:
    - O(1) (버블 정렬은 추가적인 메모리를 거의 사용하지 않는다)

* 알고리즘의 안정성:
    - 안정 정렬(stable sort): 버블 정렬은 안정 정렬이다. 동일한 값이 있을 경우, 원래 순서가 바뀌지 않는다. 즉, 동일한 값을 가진 두 원소가 있으면, 정렬 후에도 첫 번째 원소가 두 번째 원소 앞에 오게 된다.

* 최적의 알고리즘:
    - 최적화: 이미 정렬된 데이터를 효율적으로 처리하기 위해 "교환이 없으면 종료" 조건을 추가할 수 있다. 이렇게 하면 불필요한 반복을 줄일 수 있다.

### 버블 정렬 기본 논리 흐름 및 필요한 함수/논리
1. 기본 논리 흐름
    - 비교 및 교환: 리스트의 첫 번째 원소부터 마지막 원소까지, 인접한 두 원소를 비교하여 잘못된 순서라면 교환한다.
    - 반복: 매 반복마다 가장 큰 값이 리스트의 끝으로 "버블"처럼 올라가며, 남은 원소들에 대해서 같은 작업을 반복한다.
    - 최적화: 더 이상 교환이 없으면 정렬이 완료된 것이므로, 반복을 종료한다.

2. 필요한 함수/논리
* Bubble_Sort(test_list):

    리스트를 반복하여 비교하고 교환하여 정렬한다.
    비교 후 교환이 이루어지지 않으면 종료한다.

* swap(i, j) (선택사항):

    두 원소의 위치를 교환한다.
    필요시 교환을 직접 처리하는 함수로 구현할 수 있다.

* 최적화 (선택사항):

    리스트를 순회하면서 더 이상 교환이 일어나지 않으면 반복을 중지한다.

```python
# 구현 알고리즘 
def Bubble_Sort(test_list):
    n = len(test_list)
    
    # 리스트를 반복하여 비교하고 교환
    for i in range(n):
        swapped = False  # 교환 여부를 추적하는 변수
        
        # n-i-1까지 반복하는 이유는 매 반복마다 마지막 값은 정렬되기 때문
        for j in range(0, n-i-1):
            if test_list[j] > test_list[j+1]:  # 두 값을 비교
                # 값을 교환
                test_list[j], test_list[j+1] = test_list[j+1], test_list[j]
                swapped = True
        
        # 한 번의 패스에서 교환이 없으면 이미 정렬된 상태이므로 종료
        if not swapped:
            break

    return test_list

# 테스트
test_list = [64, 34, 25, 12, 22, 11, 90]
print("정렬된 리스트:", Bubble_Sort(test_list))
```

### 2. 선택 정렬 (Selection Sort)
* 기본 아이디어: 선택 정렬은 리스트에서 가장 작은(혹은 가장 큰) 원소를 찾아 첫 번째 원소와 교환하는 방식으로 리스트를 정렬하는 알고리즘이다. 이 과정은 리스트를 처음부터 끝까지 반복하며, 각 반복마다 정렬된 부분을 확장해 나간다.

* 단계:

    - 첫 번째 패스: 리스트에서 가장 작은 원소를 찾아 첫 번째 원소와 교환한다.
    - 두 번째 패스: 첫 번째 원소는 이미 정렬되었으므로, 두 번째 원소부터 끝까지 가장 작은 원소를 찾아 두 번째 원소와 교환한다.
    - 반복: 매번 나머지 원소들 중 가장 작은 값을 찾아 교환하는 과정을 리스트의 끝까지 반복한다. 이 과정은 리스트의 크기가 1이 될 때까지 진행된다.

* 장점:

    - 단순함: 선택 정렬은 알고리즘이 직관적이고 구현이 매우 간단하다.
    - 메모리 사용 적음: 선택 정렬은 in-place 정렬 알고리즘이므로, 추가적인 메모리를 사용하지 않는다.

* 단점:

    - 비효율적: 시간 복잡도가 O(n²)로, 정렬해야 할 데이터가 많을수록 비효율적이다. 각 패스마다 리스트의 나머지 부분을 모두 확인해야 하므로 속도가 느리다.
    최악의 경우에도 성능이 변하지 않기 때문에 비효율적이다. 이미 정렬된 리스트에도 동일한 시간이 소요된다.
    
* 시간 복잡도:

    - 최선: O(n²) (리스트가 이미 정렬되어 있어도, 매번 최소값을 찾기 위해 모두 비교한다)
    - 평균: O(n²) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n²) (리스트가 역순일 경우)
    
* 공간 복잡도:

    - O(1) (선택 정렬은 추가적인 메모리를 거의 사용하지 않는다)

* 알고리즘의 안정성:

    - 비안정 정렬(unstable sort): 선택 정렬은 안정 정렬이 아니다. 동일한 값이 있을 경우, 원래의 순서가 변경될 수 있다. 예를 들어, 동일한 값이 리스트에 여러 개 있을 때, 첫 번째와 두 번째 값이 뒤바뀔 수 있다.

* 최적의 알고리즘:

    - 최적화: 선택 정렬은 이미 최적화된 방식으로 작동한다. 교환 횟수를 줄이는 등의 최적화가 필요 없으며, 비교 횟수는 동일하게 O(n²)이다. 단, 선택 정렬은 다른 알고리즘에 비해 단순하며 메모리 사용량이 적기 때문에 작은 데이터셋에 유리하다.

### 선택 정렬 기본 논리 흐름 및 필요한 함수/논리
* 기본 논리 흐름

    - 비교 및 선택: 리스트에서 가장 작은(또는 가장 큰) 값을 찾아 현재 인덱스와 교환한다. 교환 대상은 매 반복마다 리스트의 나머지 원소들 중에서 선택된다.
    - 반복: 리스트의 첫 번째부터 끝까지 순차적으로 반복하면서, 각 패스마다 가장 작은 값(혹은 가장 큰 값)을 선택하여 정렬된 부분을 확장한다.
    - 최적화: 선택 정렬 자체는 알고리즘 최적화가 어려운 구조지만, 반복 횟수를 줄이기 위해 이미 정렬된 부분에 대해 비교만 수행하는 방식으로 최적화할 수 있다.

* 필요한 함수/논리

    - Selection_Sort(test_list):

        리스트에서 가장 작은 값(혹은 큰 값)을 찾아 현재 위치에 교환하며, 매 패스마다 정렬된 리스트의 크기를 확장한다.
        전체 리스트를 순회하면서 각 패스마다 가장 작은 원소를 찾고 그 원소를 현재 인덱스와 교환한다.

    - swap(i, j) (선택사항):

        두 원소 i와 j를 교환하는 함수. 선택 정렬에서 필요에 따라 직접 구현할 수 있다.
        인덱스를 받아서 해당 원소들을 리스트에서 교환한다.

    - 최적화 (선택사항):

        이미 정렬된 상태에서 불필요한 비교를 하지 않기 위해, 이미 정렬된 부분을 제외하고 나머지 부분만 처리하는 방식으로 구현할 수 있다. 그러나 선택 정렬은 기본적으로 최적화가 어려운 알고리즘이므로 최적화의 효과가 미미하다.

```python
def Selection_Sort(test_list):
    n = len(test_list)
    
    # 리스트 전체를 순회하면서
    for i in range(n):
        min_index = i  # 현재 인덱스부터 시작
        
        # i 이후의 원소들 중 가장 작은 값을 찾음
        for j in range(i + 1, n):
            if test_list[j] < test_list[min_index]:
                min_index = j
        
        # 찾은 최소값을 현재 위치와 교환
        if min_index != i:
            swap(test_list, i, min_index)
        
        print(f"현재 상태: {test_list}")
    
    return test_list

# 두 원소의 위치를 교환하는 함수
def swap(test_list, i, j):
    test_list[i], test_list[j] = test_list[j], test_list[i]

# 예시
test_list = [64, 25, 12, 22, 11]
print("정렬 전:", test_list)
sorted_list = Selection_Sort(test_list)
print("정렬 후:", sorted_list)
```

### 3. 삽입 정렬 (Insertion Sort)
* 기본 아이디어:
삽입 정렬은 리스트의 두 번째 원소부터 시작하여, 각 원소를 그 앞에 있는 원소들과 비교해 올바른 위치에 삽입하는 방식으로 리스트를 정렬하는 알고리즘이다. 이 과정은 리스트를 순차적으로 반복하며 정렬된 부분을 확장해 나간다. 마치 카드 정리와 유사한 방식으로, 리스트의 정렬된 부분에 하나씩 값을 삽입하며 정렬된다.

* 단계:

    - 첫 번째 패스: 리스트의 두 번째 원소부터 시작하여, 그 원소를 정렬된 부분에 적절한 위치에 삽입한다.
    - 두 번째 패스: 세 번째 원소를 정렬된 부분에 삽입한다. 그 후 네 번째 원소를 정렬된 부분에 삽입하는 방식으로 진행된다.
    - 반복: 매번 새 원소를 그 앞에 있는 원소들과 비교하여 올바른 위치에 삽입하는 과정을 리스트의 끝까지 반복한다.

* 장점:

    - 단순함: 삽입 정렬은 알고리즘이 직관적이고 구현이 간단하다.
    - 최선의 경우 성능: 이미 정렬된 리스트에 대해서는 비교를 최소화하고 한 번만 삽입하는 방식으로 빠르게 처리된다. 최선의 경우 시간 복잡도는 O(n)이다.
    - 알고리즘의 안정성: 삽입 정렬은 안정 정렬로, 동일한 값을 가진 원소들이 원래의 순서를 유지한다.

* 단점:

    - 비효율적: 시간 복잡도가 최악의 경우 O(n²)로, 리스트의 크기가 커지면 성능이 급격히 떨어진다. 특히 역순으로 정렬된 리스트에서 매우 비효율적이다.
    - 큰 데이터셋에 비효율적: 삽입 정렬은 작은 데이터셋에서 유리하지만, 큰 데이터셋에서는 다른 효율적인 정렬 알고리즘에 비해 성능이 떨어진다.

* 시간 복잡도:

    - 최선: O(n) (리스트가 이미 정렬된 경우, 각 원소를 한 번만 비교하고 삽입하므로 시간 복잡도가 O(n)이다.)
    - 평균: O(n²) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n²) (리스트가 역순으로 정렬된 경우)

* 공간 복잡도:

    - O(1) (삽입 정렬은 추가적인 메모리를 거의 사용하지 않으며, in-place 방식으로 작동한다.)

* 알고리즘의 안정성:

    - 안정 정렬(stable sort): 삽입 정렬은 안정 정렬이다. 동일한 값이 있을 경우, 원래의 순서가 바뀌지 않는다. 즉, 동일한 값을 가진 두 원소가 있을 경우, 정렬 후에도 첫 번째 원소가 두 번째 원소 앞에 오게 된다.

* 최적의 알고리즘:

    - 최적화: 이미 정렬된 리스트에 대해 최적화된 성능을 보이므로, 작은 데이터셋이나 거의 정렬된 리스트에 유리하다. 또한, 비교와 이동을 최소화하는 방식을 추가적으로 최적화할 수 있다.

### 삽입 정렬 기본 논리 흐름 및 필요한 함수/논리
* 기본 논리 흐름

    - 비교 및 삽입: 리스트에서 두 번째 원소부터 시작하여, 각 원소를 그 앞에 있는 원소들과 비교한다. 만약 현재 원소가 그 앞의 원소보다 작으면, 그 원소들을 오른쪽으로 밀어내고, 현재 원소를 올바른 위치에 삽입한다.
    - 반복: 리스트의 모든 원소를 순차적으로 처리하며, 매번 삽입이 이루어진 후 정렬된 부분을 확장한다.
    - 최적화: 이미 정렬된 상태에서는 비교와 삽입 작업을 최소화할 수 있다. 최선의 경우 시간 복잡도는 O(n)이다.

* 필요한 함수/논리

    - Insertion_Sort(test_list): 리스트의 두 번째 원소부터 시작하여, 각 원소를 정렬된 부분에 삽입하며 리스트를 정렬한다. 매번 새로운 원소를 삽입할 때, 그 원소가 들어갈 올바른 위치를 찾아 삽입하는 작업을 수행한다.

    - key (선택사항): 삽입 정렬에서 삽입할 원소를 변수에 저장하여, 그 원소를 비교하고 올바른 위치를 찾아 삽입하는 작업을 한다.

    - 최적화 (선택사항): 이미 정렬된 부분에 대해서는 추가적인 비교를 하지 않도록 하여 성능을 향상시킬 수 있다. 이를 통해 최선의 경우 O(n) 시간이 걸리며, 이미 정렬된 리스트에서는 최적의 성능을 발휘한다.

```python
def Insertion_Sort(test_list):
    # 리스트의 두 번째 원소부터 시작
    for i in range(1, len(test_list)):
        key = test_list[i]  # 현재 비교할 원소
        j = i - 1  # 비교할 원소의 바로 앞 원소 인덱스
        
        # key가 정렬된 부분의 원소들보다 작다면, 그 원소들을 오른쪽으로 밀어내고 올바른 위치를 찾는다.
        while j >= 0 and test_list[j] > key:
            test_list[j + 1] = test_list[j]
            j -= 1
        
        # key를 정렬된 부분의 올바른 위치에 삽입
        test_list[j + 1] = key
        
        # 중간 결과 출력 (선택적)
        print(f"현재 상태: {test_list}")
    
    return test_list

# 예시
test_list = [64, 25, 12, 22, 11]
print("정렬 전:", test_list)
sorted_list = Insertion_Sort(test_list)
print("정렬 후:", sorted_list)
```

### 4. 퀵 정렬 (Quick Sort)
* 기본 아이디어: 퀵 정렬은 분할 정복(divide-and-conquer) 알고리즘으로, 리스트를 두 개의 하위 리스트로 나누고, 각 하위 리스트를 독립적으로 정렬한 뒤, 병합하는 방식으로 동작한다. 퀵 정렬은 리스트에서 하나의 원소를 '피벗(pivot)'으로 선택하고, 이를 기준으로 작은 값들은 왼쪽, 큰 값들은 오른쪽에 배치한 후, 다시 재귀적으로 같은 작업을 수행하여 리스트를 정렬한다.

* 단계:

    - 피벗 선택: 리스트에서 하나의 원소를 피벗으로 선택한다. 보통 첫 번째, 마지막 원소, 중간 원소 중 하나를 선택할 수 있다.
    - 분할: 피벗을 기준으로, 피벗보다 작은 값들은 왼쪽, 큰 값들은 오른쪽으로 배치한다. 이때 피벗은 이미 정렬된 상태가 된다.
    - 재귀 호출: 왼쪽과 오른쪽의 하위 리스트에 대해 다시 같은 작업을 재귀적으로 수행한다. 리스트가 더 이상 두 원소 이하로 나눠지지 않으면 정렬이 완료된다.

* 장점:

    - 효율성: 퀵 정렬은 평균적으로 O(n log n) 시간 복잡도를 가지고 있어, 다른 많은 정렬 알고리즘에 비해 매우 빠르다.
    - 인-플레이스 정렬: 퀵 정렬은 추가적인 메모리 공간을 거의 사용하지 않고, 원본 리스트에서 정렬을 수행한다. 따라서 공간 복잡도가 O(log n) 정도로 매우 적다.
    - 분할 정복: 재귀적으로 하위 문제로 나누어 해결하므로 분할 정복 전략에 의해 문제가 효율적으로 해결된다.

* 단점:

    - 최악의 경우 성능: 피벗 선택이 잘못되면, 예를 들어 리스트가 이미 정렬된 상태일 경우 최악의 시간 복잡도인 O(n²)까지 성능이 떨어진다.
    - 불안정 정렬: 퀵 정렬은 안정 정렬이 아니므로, 동일한 값이 있을 경우 순서가 바뀔 수 있다.

* 시간 복잡도:

    - 최선: O(n log n) (피벗 선택이 잘 되어 분할이 균등하게 이루어지는 경우)
    - 평균: O(n log n) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n²) (리스트가 역순으로 정렬된 경우 혹은 피벗 선택이 잘못된 경우)

* 공간 복잡도:

    - O(log n) (재귀 호출에 의한 스택 공간을 포함한 공간 복잡도)
        퀵 정렬은 기본적으로 인-플레이스 정렬이지만, 재귀 호출 스택 때문에 추가적인 메모리 공간을 조금 사용한다.

* 알고리즘의 안정성:

- 불안정 정렬(unstable sort): 퀵 정렬은 안정 정렬이 아니다. 동일한 값이 있을 경우, 원래의 순서가 바뀔 수 있다. 예를 들어, 동일한 값의 두 원소가 정렬 후 위치가 바뀌게 될 수 있다.
최적의 알고리즘:

* 피벗 선택 최적화: 퀵 정렬의 성능을 향상시키기 위해 피벗을 랜덤으로 선택하거나, 리스트의 첫 번째, 마지막, 중간 값을 비교하여 더 나은 피벗을 선택하는 전략을 사용할 수 있다.
힙 정렬과 병합 정렬: 최악의 경우 성능이 중요한 상황에서는 퀵 정렬 대신 병합 정렬이나 힙 정렬을 사용할 수 있다. 하지만 평균적으로는 퀵 정렬이 더 빠르다.


### 퀵 정렬 기본 논리 흐름 및 필요한 함수/논리
* 기본 논리 흐름

    - 피벗 선택 및 분할: 리스트에서 하나의 원소를 선택하여, 이를 기준으로 리스트를 두 부분으로 나눈다. 피벗보다 작은 원소들은 왼쪽, 큰 원소들은 오른쪽에 배치한다.
    - 재귀적 정렬: 피벗을 기준으로 나눈 왼쪽과 오른쪽 부분에 대해 각각 퀵 정렬을 재귀적으로 적용한다. 재귀 호출이 끝나면, 리스트는 정렬된다.
    - 반복: 이 과정을 계속해서 반복하여 리스트 전체가 정렬된다.
    - 최적화: 최악의 경우(예: 피벗이 항상 가장 큰/작은 원소인 경우) 성능이 O(n²)로 떨어지므로, 피벗을 무작위로 선택하거나 "median-of-three" 방식(첫 번째, 마지막, 중간 원소 중 중앙값을 피벗으로 선택)으로 피벗 선택을 최적화할 수 있다.

* 필요한 함수/논리

    - Quick_Sort(test_list): 주어진 리스트에서 피벗을 선택하고, 해당 피벗을 기준으로 리스트를 나누어 왼쪽과 오른쪽 부분에 대해 각각 퀵 정렬을 재귀적으로 수행한다.

    - Partition(start, end): 리스트를 피벗을 기준으로 분할하는 함수. 피벗을 기준으로 왼쪽에는 피벗보다 작은 값들, 오른쪽에는 피벗보다 큰 값들이 오도록 배치하고, 피벗이 최종 위치로 정렬된다.

    - 재귀적 호출: 퀵 정렬은 분할된 각 리스트에 대해 재귀적으로 호출되며, 리스트가 한 원소 이하로 분할되면 정렬이 완료된다.



```python
def quick_sort(arr):
    if len(arr) <= 1:  # 리스트의 길이가 1 이하일 경우 이미 정렬됨
        return arr
    pivot = arr[len(arr) // 2]  # 피벗을 리스트의 중간 원소로 선택
    left = [x for x in arr if x < pivot]  # 피벗보다 작은 원소들
    middle = [x for x in arr if x == pivot]  # 피벗과 같은 원소들
    right = [x for x in arr if x > pivot]  # 피벗보다 큰 원소들
    # 재귀적으로 왼쪽과 오른쪽 리스트를 정렬하고, 병합
    return quick_sort(left) + middle + quick_sort(right)

# 예시
test_list = [3, 6, 8, 10, 1, 2, 1]
print("정렬 전:", test_list)
sorted_list = quick_sort(test_list)
print("정렬 후:", sorted_list)
```

### 5. 병합 정렬 (Merge Sort)
* 기본 아이디어:
병합 정렬은 분할 정복(divide-and-conquer) 알고리즘으로, 리스트를 재귀적으로 절반씩 나누고, 나눠진 리스트를 각각 정렬한 후, 정렬된 리스트들을 병합하여 최종적으로 정렬된 리스트를 만드는 방식이다. 병합 정렬은 비교적 안정적인 정렬 알고리즘이며, 항상 일정한 성능을 보장한다.

* 단계:

    - 분할: 주어진 리스트를 반으로 나눈다. 이 과정을 리스트가 하나의 원소가 될 때까지 반복한다.
    - 정렬 및 병합: 각 나뉜 리스트가 하나의 원소로 나뉘면, 이를 정렬하고 두 개의 정렬된 리스트를 병합하여 더 큰 정렬된 리스트를 만든다.
    - 재귀 호출: 리스트의 크기가 1이 될 때까지 재귀적으로 리스트를 나누고 병합하는 작업을 반복한다.

* 장점:

    - 효율성: 병합 정렬은 항상 O(n log n) 시간 복잡도를 보장한다. 이는 최악의 경우에도 동일하다.
    - 안정성: 병합 정렬은 안정 정렬이다. 즉, 동일한 값들이 있을 경우 원래의 순서를 유지한다.
    - 분할 정복: 리스트를 반으로 나누어 정렬하고, 두 개씩 병합하는 방식으로 동작하여 매우 체계적이다.

* 단점:

    - 공간 복잡도: 병합 정렬은 O(n)의 추가 공간을 필요로 한다. 원본 리스트를 나누고 병합하기 위해 추가적인 배열이 필요하다.
    - 비교적 느림: 평균적으로는 퀵 정렬보다 속도가 느릴 수 있다.

* 시간 복잡도:

    - 최선: O(n log n) (리스트가 이미 정렬되어 있어도 항상 동일한 성능을 보인다)
    - 평균: O(n log n) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n log n) (리스트가 역순으로 정렬되어 있어도 동일하게 O(n log n) 성능을 보인다)

* 공간 복잡도:

    - O(n) (병합 정렬은 추가적인 배열을 필요로 하며, 이로 인해 O(n) 만큼의 공간을 사용한다.)

* 알고리즘의 안정성:

    - 안정 정렬(stable sort): 병합 정렬은 안정 정렬이다. 동일한 값들이 있을 경우, 원래의 순서가 바뀌지 않는다.

* 최적의 알고리즘:

    병합 정렬은 최악의 경우에도 항상 O(n log n)의 성능을 보장하므로, 일정한 성능이 중요한 상황에서 매우 유용하다. 그러나 추가적인 메모리 공간을 필요로 하므로, 메모리 사용이 중요한 경우에는 퀵 정렬이나 힙 정렬을 고려할 수 있다.


### 병합 정렬 기본 논리 흐름 및 필요한 함수/논리
* 기본 논리 흐름:

    - 분할: 리스트를 반으로 나누고, 각 하위 리스트가 하나의 원소가 될 때까지 분할한다.
    - 병합: 두 개의 정렬된 하위 리스트를 하나의 정렬된 리스트로 병합한다. 병합은 두 리스트에서 가장 작은 원소를 하나씩 비교하여 하나씩 정렬된 리스트에 삽입한다.
    - 재귀 호출: 리스트가 하나의 원소로 나뉘면, 이를 병합하는 작업을 재귀적으로 계속해서 반복한다.

* 필요한 함수/논리:

    - Merge_Sort(test_list): 주어진 리스트를 재귀적으로 나누어 정렬한 후, 병합한다.
    - Merge(left, right): 두 개의 정렬된 리스트를 병합하는 함수. 이 함수는 두 리스트를 비교하여 더 작은 값을 먼저 결과 리스트에 삽입하며 병합한다.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    # 리스트를 반으로 나누기
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])  # 왼쪽 절반
    right = merge_sort(arr[mid:])  # 오른쪽 절반
    
    # 두 리스트를 병합
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    # 두 리스트의 원소를 하나씩 비교하여 작은 값을 결과에 추가
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # 남은 원소들 추가
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

# 예시
test_list = [3, 6, 8, 10, 1, 2, 1]
print("정렬 전:", test_list)
sorted_list = merge_sort(test_list)
print("정렬 후:", sorted_list)
```

### 6. 힙 정렬 (Heap Sort)
* 기본 아이디어: 힙 정렬은 힙 자료구조를 기반으로 한 정렬 알고리즘이다. 힙은 이진 트리 형태로 구성되며, 최대 힙(Max Heap) 또는 최소 힙(Min Heap) 구조를 가진다. 힙 정렬은 먼저 주어진 리스트를 힙으로 변환한 후, 힙의 루트에 있는 최대 값(혹은 최소 값)을 마지막에 배치하고, 나머지 원소에 대해서도 반복적으로 힙 구조를 유지하며 정렬한다.

* 단계:

    - 힙 만들기: 주어진 리스트를 힙 자료구조로 변환한다. 보통 최대 힙을 사용하여, 힙의 루트에 가장 큰 값이 오도록 만든다.
    - 정렬 반복: 힙의 루트에 있는 최대 값(혹은 최소 값)을 배열의 마지막 원소와 교환하고, 남은 부분에서 다시 힙 구조를 유지하며, 위 과정을 반복한다.
    - 힙 구조 재정비: 각 교환 후, 힙에서 빠져나간 요소를 제외한 나머지 원소들로 새로운 힙을 형성한다.

* 장점:

    - 효율성: 힙 정렬은 평균적으로 O(n log n)의 시간 복잡도를 가지고 있다. 이로 인해 퀵 정렬이나 병합 정렬과 같은 다른 O(n log n) 알고리즘과 비슷한 성능을 보인다.
    - 인-플레이스 정렬: 힙 정렬은 원본 배열에서 정렬을 수행하므로, 추가적인 메모리 공간을 거의 사용하지 않는다. 공간 복잡도가 O(1)로 매우 효율적이다.
    - 일관된 성능: 퀵 정렬과 달리 힙 정렬은 최악의 경우에도 O(n log n)의 성능을 보장한다.

* 단점:

    - 불안정 정렬: 힙 정렬은 안정 정렬이 아니다. 동일한 값이 있을 경우, 원래의 순서가 바뀔 수 있다.
    - 비교적 느림: 힙 정렬은 퀵 정렬이나 병합 정렬에 비해 실제 동작 속도가 느릴 수 있다. 힙 자료구조의 특성상, 병합 정렬이나 퀵 정렬보다 많은 비교와 교환이 발생할 수 있다.

* 시간 복잡도:

    - 최선: O(n log n) (항상 일정한 성능을 보인다)
    - 평균: O(n log n) (리스트가 무작위로 주어졌을 경우)
    - 최악: O(n log n) (리스트가 역순으로 정렬되어 있어도 동일하게 O(n log n))

* 공간 복잡도:

    -O(1) (인-플레이스 정렬이므로 추가적인 메모리 공간을 거의 사용하지 않는다.)

* 알고리즘의 안정성:

    - 불안정 정렬(unstable sort): 힙 정렬은 안정 정렬이 아니다. 동일한 값의 원소가 있을 경우, 정렬 후 순서가 바뀔 수 있다.

* 최적의 알고리즘:

    힙 정렬은 퀵 정렬의 최악의 경우 성능 문제를 해결하면서도 O(n log n) 성능을 보장한다. 그러나 실제로는 퀵 정렬이나 병합 정렬에 비해 상대적으로 느릴 수 있다. 따라서 일반적으로 퀵 정렬이나 병합 정렬이 더 빠르지만, 힙 정렬은 최악의 경우 성능이 일정하므로 안정적인 성능을 요구하는 환경에서 유용하다.

### 힙 정렬 기본 논리 흐름 및 필요한 함수/논리
* 기본 논리 흐름:

    - 힙 만들기: 주어진 리스트를 힙 자료구조로 변환한다. 보통 최대 힙을 사용하여, 힙의 루트에 가장 큰 값이 오도록 만든다.
    - 최대값 추출: 힙에서 루트 원소(가장 큰 값)를 배열의 마지막 원소와 교환한 후, 나머지 원소들로 다시 힙을 구성한다.
    - 반복: 힙 구조를 재정비하면서, 힙에서 최대값을 추출하고 교환하는 작업을 리스트가 정렬될 때까지 반복한다.

* 필요한 함수/논리:

    - Heapify(arr, n, i): 주어진 리스트에서 인덱스 i를 루트로 하는 서브트리를 힙 구조로 변환하는 함수. 이는 힙의 속성을 유지하는 작업을 한다.
    - Heap_Sort(arr): 리스트를 힙 자료구조로 변환한 후, 최대값을 추출하여 정렬하는 함수.


```python
def heapify(arr, n, i):
    largest = i  # 루트
    left = 2 * i + 1  # 왼쪽 자식
    right = 2 * i + 2  # 오른쪽 자식

    # 왼쪽 자식이 더 크면 largest 업데이트
    if left < n and arr[left] > arr[largest]:
        largest = left

    # 오른쪽 자식이 더 크면 largest 업데이트
    if right < n and arr[right] > arr[largest]:
        largest = right

    # largest가 루트가 아니라면 swap하고, 그 자식에서 다시 heapify
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # swap
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    # 힙을 만들기 (배열을 최대 힙으로 변환)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # 최대 힙에서 루트를 빼고, 그 원소를 마지막 원소와 교환한 후 다시 heapify
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # swap
        heapify(arr, i, 0)

# 예시
test_list = [3, 6, 8, 10, 1, 2, 1]
print("정렬 전:", test_list)
heap_sort(test_list)
print("정렬 후:", test_list)
```

#### 1. 버블 정렬 (Bubble Sort)
* 문제: 주어진 리스트를 오름차순으로 정렬하시오. 버블 정렬 알고리즘을 구현하여 리스트를 정렬하는 방법을 사용하시오. 정렬하는 과정에서 리스트의 각 단계도 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 각 패스마다 리스트가 어떻게 변화하는지 출력합니다.

* 입력 예시:

    - [5, 2, 9, 1, 5, 6]


* 출력 예시 

    - 패스 1: [2, 5, 1, 5, 6, 9]

    - 패스 2: [2, 1, 5, 5, 6, 9]

    - 패스 3: [1, 2, 5, 5, 6, 9]

    - 정렬된 리스트: [1, 2, 5, 5, 6, 9]

```python
# 해당 코드의 중요한점은 j의 반복이 중요
# 1차원 배열에서 0번째 인덱스가 끝까지 비교해서 (0,1),(1,2),...,(4,5) 비교하면서 전부 비교후 가장 큰값은 오른쪽으로 이동 다 함
# j의 반복이 진행되는 동안 계속해서 배열의 길이만큼 반복할 필요가 없음
# j의 반복을 줄여주는게 효율적

def bb_sort(arr):
    length = len(arr)   # 0~n번까지 반복을 위한 변수 설정
    for i in range(length-1):
        swapped = False
        for j in range(length-1-i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        print(f'패스 {i+1} : {arr}')
        if not swapped:
            break
    print(f'정렬된 리스트: {arr}')
    
bb_sort([5, 2, 9, 1, 5, 6])
                
```

```python
# 개인이 구현해본 알고리즘_직접 만들어 보세요!
def Bubble_Sort_YJ (test_list):

    print(f"원본 정렬: {test_list}")

    for i in range(len(test_list) - 1):
        # i 와 i + 1번째의 요소를 비교. 
        for j in range(len(test_list) - 1):

            if min(test_list[j], test_list[j + 1]) == test_list[j + 1]:
                # 비교했는데 뒤의 것이 더 작다면 -> 교환
                i_value = test_list[j + 1]
                i_plus_1_value = test_list[j]

                test_list[j] = i_value
                test_list[j + 1] = i_plus_1_value

            #print(f'{j + 1}번째 요소 정렬 : {test_list}')
            
        #print(f"{i + 1}번째 정렬 : {test_list}")

    return test_list

test_list = [5, 2, 9, 1, 5, 6]
print(f"버블 정렬: {Bubble_Sort_YJ(test_list)}")
```

#### 2. 선택 정렬 (Selection Sort)
* 문제: 주어진 리스트에서 가장 작은 값부터 차례대로 찾아서 정렬하시오. 선택 정렬을 구현한 후, 정렬이 끝날 때까지 각 단계에서 리스트 상태를 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 각 단계에서 리스트가 어떻게 변화하는지 출력합니다.

* 입력 예시:

    - [64, 25, 12, 22, 11]

* 출력 예시:
    
    - 패스 1: [11, 25, 12, 22, 64]
    - 패스 2: [11, 12, 25, 22, 64]
    - 패스 3: [11, 12, 22, 25, 64]
    - 패스 4: [11, 12, 22, 25, 64]
    - 정렬된 리스트: [11, 12, 22, 25, 64]
    

```python
# 내부 반복을 통해서 최소의 값이 순차적으로 왼쪽정렬
# 최솟값을 왼쪽으로 이동시키기 위해 i와 j의 값의 위치를 바꿈

def ss_sort(arr):
    length = len(arr)
    for i in range(length-1):
        min_index = i
        for j in range(i+1, length):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
        print(f'패스 {i+1} : {arr}')
    print(f'정렬된 리스트: {arr}')
    
ss_sort([64, 25, 12, 22, 11])
```

#### 3. 삽입 정렬 (Insertion Sort)
* 문제: 삽입 정렬 알고리즘을 사용하여 리스트를 오름차순으로 정렬하시오. 각 삽입 단계를 출력하고, 최종적으로 정렬된 리스트를 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 각 삽입이 이루어질 때마다 리스트 상태를 출력합니다.

* 입력 예시:

    - [12, 11, 13, 5, 6]

* 출력 예시:

    - 삽입 1: [11, 12, 13, 5, 6]
    - 삽입 2: [11, 12, 13, 5, 6]
    - 삽입 3: [5, 11, 12, 13, 6]
    - 삽입 4: [5, 6, 11, 12, 13]
    - 정렬된 리스트: [5, 6, 11, 12, 13]

```python
def i_sort_error(arr):
    length = len(arr)
    for i in range(1, length):
        insert_num = arr[i]
        j = i - 1
        for j in range(i - 1, -1, -1):
            if arr[j] > insert_num:
                arr[j + 1] = arr[j]
            else:
                arr[j + 1] = insert_num
                break
        print(f'삽입 {i}: {arr}')
    print(f'정렬된 리스트: {arr}')

def i_sort(arr):
    length = len(arr)
    for i in range(1, length):
        insert_num = arr[i]
        j = i - 1
        # j를 중복 정의하지 않도록 하고, arr[j]와 insert_num을 비교
        while j >= 0 and arr[j] > insert_num:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = insert_num
        print(f'삽입 {i}: {arr}')
    print(f'정렬된 리스트: {arr}')

i_sort([12, 11, 13, 5, 6])
```

```python
# 예시 문제 풀이
def Insertion_Sort_YJ(test_list):

    print(f"원본 정렬: {test_list}")

    sorted_list = [test_list[0]]
    unsorted_list = test_list[1: ] 
    

    for i in range(len(test_list)):
        unsorted_list = test_list[i + 1: ]

        # print(f"{sorted_list}, {unsorted_list}")

        if len(unsorted_list) == 0:
            break

        insert_value = unsorted_list[0]

        if insert_value <= sorted_list[0]:
         # sorted_list 의 앞에 추가 -> 하나씩 밀어버려 
         sorted_list = [unsorted_list[0]] + sorted_list

        elif insert_value > sorted_list[-1]:
        # sorted_list 의 뒤에 추가
            sorted_list.append(unsorted_list[0])

        else: 
        # 위치를 찾아내
            for j in range(len(sorted_list)):
                if sorted_list[j] > insert_value:
                    sorted_list = sorted_list[:j] + [insert_value] + sorted_list[j:]
                    break
        
    return sorted_list

test_list = [12, 11, 13, 5, 6]
print(f"삽입 정렬: {Insertion_Sort_YJ(test_list)}")
```

#### 4. 퀵 정렬 (Quick Sort)
* 문제: 주어진 리스트를 퀵 정렬 알고리즘을 이용하여 오름차순으로 정렬하시오. 피벗을 기준으로 나누는 과정을 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 피벗을 기준으로 리스트가 분할될 때마다 상태를 출력합니다.

* 입력 예시:

    - [10, 7, 8, 9, 1, 5]

* 출력 예시:

    - 피벗: 10, 분할 후: [1, 5, 7, 8, 9, 10]
    - 피벗: 5, 분할 후: [1, 5, 7, 8, 9]
    - 피벗: 7, 분할 후: [1, 5, 7, 8, 9]
    - 정렬된 리스트: [1, 5, 7, 8, 9, 10]

```python
# 전 재귀로 품
# 재귀 함수를 통해 피벗을 정하고 피벗 기준으로 왼쪽 오른쪽나눔
# 왼쪽과 오른쪽으로 나뉜 함수에서 또 피벗을 정해 정렬
def q_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left_list = []
    right_list = []

    for num in arr[1:]:
        if num < pivot:
            left_list.append(num)
        else:
            right_list.append(num)

    print(f'피벗: {pivot}, 분할 후: {left_list + [pivot] + right_list}')

    return q_sort(left_list) + [pivot] + q_sort(right_list)

arr = [10, 7, 8, 9, 1, 5]
sorted_arr = q_sort(arr)
print(f'정렬된 리스트: {sorted_arr}')
```

```python
# 예시 문제 풀이 
def Quick_Sort_YJ(test_list):
    # print(f"원본 리스트: {test_list}")
    
    # 기본적인 퀵 정렬 조건: 리스트가 길이가 1 이하일 경우 이미 정렬된 상태
    if len(test_list) <= 1:
        # print(f"리스트 길이가 1 이하이므로 정렬 완료: {test_list}")
        return test_list
    
    # 기준값(pivot) 설정: 마지막 원소를 기준값으로 설정
    pivot = test_list.pop()
    # print(f"기준값 (pivot): {pivot}")
    
    # 기준값을 기준으로 리스트를 나누기
    left_list, right_list = Parted_list(pivot, test_list)

    # print(f"분할 후 좌측 리스트: {left_list}, 우측 리스트: {right_list}")

    # 재귀적으로 좌측과 우측 리스트를 정렬
    sorted_left_list = Quick_Sort_YJ(left_list)
    sorted_right_list = Quick_Sort_YJ(right_list)

    # print(f"정렬된 좌측 리스트: {sorted_left_list}, 정렬된 우측 리스트: {sorted_right_list}")

    # 정렬된 좌측, 우측 리스트와 pivot을 합쳐서 반환
    sorted_list = sorted_left_list + [pivot] + sorted_right_list
    print(f"최종 정렬 결과: {sorted_list}")
   
    return sorted_list


def Parted_list(pivot, parted_list):
    left_list = []
    right_list = []

    # pivot을 기준으로 리스트 분할
    for item in parted_list:
        if item <= pivot:
            left_list.append(item)
        else:
            right_list.append(item)

    # print(f"기준값 {pivot}을 기준으로 분할된 좌측 리스트: {left_list}, 우측 리스트: {right_list}")
    return left_list, right_list

test_list = [10, 7, 8, 9, 1, 5]
print(f"퀵 정렬: {Quick_Sort_YJ(test_list)}")
```

#### 5. 병합 정렬 (Merge Sort)
* 문제: 병합 정렬 알고리즘을 사용하여 주어진 리스트를 오름차순으로 정렬하시오. 병합 과정도 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 병합 과정에서 두 하위 리스트가 합쳐지는 과정을 출력합니다.

* 입력 예시:

    - [38, 27, 43, 3, 9, 82, 10]

* 출력 예시:

    - 병합: [38, 27, 43], [3, 9, 82, 10] -> [3, 9, 10, 27, 38, 43, 82]
    - 정렬된 리스트: [3, 9, 10, 27, 38, 43, 82]

```python
# 예시 문제 풀이 
def Merge_Sort_YJ(test_list):
    # 길이가 1이하일 때까지 분할하고 반환
    if len(test_list) <= 1:
        print("분할:", test_list)
        return test_list
    
    # 리스트를 두 부분으로 나누기
    left_list, right_list = Split_List(test_list)

    # 재귀적으로 병합정렬을 수행
    splited_left_list = Merge_Sort_YJ(left_list)
    splited_right_list = Merge_Sort_YJ(right_list)

    # 정렬된 두 리스트를 병합하여 반환
    return Merge_List(splited_left_list, splited_right_list)

def Split_List(test_list):
    # 리스트의 중간 인덱스를 기준으로 나누기
    length = len(test_list)
    pivot = length // 2  # 중간 인덱스를 계산

    left_list = test_list[:pivot]
    right_list = test_list[pivot:]

    print(f"좌측 정렬: {left_list}, 우측 정렬: {right_list}")

    return left_list, right_list

def Merge_List(left, right):
    merged_list = []  # 결과를 담을 새로운 리스트

    # 두 리스트를 병합할 때 비교하면서 진행
    while left and right:  # left와 right 모두 비어있지 않다면
        if left[0] <= right[0]:  # 왼쪽 리스트의 첫 번째 원소가 더 작거나 같으면
            merged_list.append(left.pop(0))  # 왼쪽 원소를 병합된 리스트에 추가
        else:
            merged_list.append(right.pop(0))  # 오른쪽 원소를 병합된 리스트에 추가
    
    # 만약 한 리스트에 원소가 남았다면 나머지 원소들을 그대로 병합
    merged_list.extend(left)  # 왼쪽 리스트에 남은 원소들
    merged_list.extend(right)  # 오른쪽 리스트에 남은 원소들

    return merged_list

test_list = [38, 27, 43, 3, 9, 82, 10]
print(f"병합 정렬: {Merge_Sort_YJ(test_list)}")
```

#### 6. 힙 정렬 (Heap Sort)

* 문제: 주어진 리스트를 힙 정렬 알고리즘을 이용하여 오름차순으로 정렬하시오. 힙 구조에서 가장 큰 원소를 추출하는 과정을 출력하시오.

* 입력: 길이가 n인 정수 리스트 arr가 주어집니다. (1 ≤ n ≤ 1000)

* 출력: 오름차순으로 정렬된 리스트를 출력하고, 각 단계에서 최대 힙에서 가장 큰 원소를 추출하는 과정을 출력합니다.

* 입력 예시:

    - [3, 19, 1, 14, 8, 7]

* 출력 예시:

    - 힙 구조: [19, 14, 8, 3, 7, 1] -> 추출된 원소: 19, 나머지 힙: [14, 8, 1, 3, 7]
    - 힙 구조: [14, 7, 8, 3, 1] -> 추출된 원소: 14, 나머지 힙: [8, 7, 1, 3]
    - 힙 구조: [8, 7, 1, 3] -> 추출된 원소: 8, 나머지 힙: [7, 3, 1]
    - 정렬된 리스트: [1, 3, 7, 8, 14, 19]

```python
# 문제 풀이
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    
    result = []
    
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        result.append(arr[i])
        heapify(arr, i, 0)
        print(f'힙 구조: {arr[:i]} -> 추출된 원소: {arr[i]}, 나머지 힙: {arr[:i]}')
    
    result.append(arr[0])
    print(f'정렬된 리스트: {result[::-1]}')

arr = [3, 19, 1, 14, 8, 7]
heap_sort(arr)
```
